```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data0 = pd.read_csv("results/classification1_2.csv")
data1 = pd.read_csv("results/classification2_1.csv")
data2 = pd.read_csv("results/classification2_0.csv")
data3 = pd.read_csv("results/classification2_3.csv")
```

```python
fig, ax = plt.subplots(2, 2)

ax[0][0].plot(data0["predict"], color="red", label="Prediction")
ax[0][0].plot(data0["target"], color="green", label="Target")
ax[0][1].plot(data1["predict"], color="red", label="Prediction")
ax[0][1].plot(data1["target"], color="green", label="Target")
ax[1][0].plot(data2["predict"], color="red", label="Prediction")
ax[1][0].plot(data2["target"], color="green", label="Target")
ax[1][1].plot(data3["predict"], color="red", label="Prediction")
ax[1][1].plot(data3["target"], color="green", label="Target")
```
