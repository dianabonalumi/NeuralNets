```python
data_class = pd.read_csv("classifier.csv")
data_class
```

```python
t = data_class["target"] == 1
p = data_class["prediction"] == 0
```

```python
TP = 0
TN = 0
FP = 0
FN = 0

for i in range(len(t)):
    if t[i] == p[i] == True:
        TP += 1
    elif t[i] == p[i] == False:
        TN += 1
    elif t[i] == True and p[i] == False:
        FN += 1
    else:
        FP += 1
```

```python
prec = TP / (TP + FP)
rec = TP / (TP + FN)

print(prec, rec)
```

```python
f1 = 2 * prec * rec / (prec + rec)
print("F1 score: ", f1)
```
